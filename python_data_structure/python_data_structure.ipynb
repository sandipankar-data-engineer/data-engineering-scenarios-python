### Adding elements to the list

* The append() method appends an element to the end of a list.
* The insert() method adds a new element at any position in the list.
* The extend() method is used to add all the items from another list.

```python
# appending "100" at the end of the list
numbers = [1, 3, 2, 7, 9, 4]
numbers.append(100)
print(numbers)
```

```python
# inserts the number "100" at index 2
numbers = [1, 3, 2, 7, 9, 4]
numbers.insert(2, 100)
print(numbers)
```

```python
# add all the items from fav_items to cart
cart = ["T-shirt", "Lamp", "Pen"]
fav_items = ["Headphones", "Phone"]
cart.extend(fav_items)
print(cart)
```

### Removing elements from a list

* del: statement allows you to remove an element from a list by specifying the position of the element.
* pop() method removes the last element from a list and returns that element.
* remove() method is used to remove the first occurance from a list.
* clear() method empties a list.

```python
# deleting 1st element from the list
numbers = [1, 3, 2, 7, 9, 4]
del numbers[0]
print(numbers)
```

```python
# popping last element out of the list
numbers = [1, 3, 2, 7, 9, 4]
last = numbers.pop()
print(last)
```

```python
# remove an element by "Pen"
cart = ["T-shirt", "Lamp", "Pen", "Book", "Pen"]
print(cart)
# Remove "Pen" from the list
cart.remove("Pen")
print(cart)
```

```python
# emptying an entire list
cart = ["T-shirt", "Lamp", "Pen", "Book", "Pen"]
print(cart)
# Remove "Pen" from the list
cart.clear()
print(cart)
```

### List Slicing

list[begin: end: step] <br>
being - starting index <br>
end - boundary index (is not included) <br>
step -  determines how many elements the slice "steps" forward or backward
<br>
<br>
image.png

```python
my_list = ['item'+str(i) for i in range(1,10)]
print(my_list)
```

```python
# First n elements
print(my_list[:3])
```

```python
# Last n elements
print(my_list[-3:])
```

```python
# Every nth element from a list
print(my_list[::2])
```

```python
# Reversing a list
print(my_list[::-1])
```

```python
# Slicing a list
print(my_list[2:6])
print(my_list[-4:-2])
print(my_list[2:8:2])
print(my_list[-1::-2])
```

### Converting to a list

```python
# Converting an iterable (such as strings, range() etc.) into a list

# Coverts a string into a list
vowels_list = list('aeiou')
print(vowels_list)

# Converts a range into a list
numbers_list = list(range(1, 6))
print(numbers_list)
```

### Functions

* append():Adds an item to the end of the list
* extend():Adds all the items from another list (or any other iterable)
* insert():Inserts an item at the specified index
* remove():Removes the specified value from the list
* pop():Returns and removes the last item or item at the given index
* clear():Removes all items from the list
* index():Returns the index of the first matched item
* count():Returns the count of the specified item
* sort():Sorts the list in ascending/descending order
* reverse():Reverses the list
* copy():Returns the shallow copy of a list
* len():Returns number of items
* sum():Sum of all the elements
* min():Min value in a list
* max():Max value in a list

```python
# append
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item6']
list1.append('item7')
print(list1)
```

```python
# extend
list1 = ['item1', 'item2', 'item3']
list2 = ['item4', 'item5', 'item6']
list1.extend(list2)
print(list1)
```

```python
# insert
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item6']
list1.insert(2,'item2.5')
print(list1)
```

```python
# remove
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item2', 'item6']
list1.remove('item2')
print(list1)
```

```python
# pop
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item6']
list1.pop()
print(list1)
```

```python
# clear
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item6']
list1.clear()
print(list1)
```

```python
# index
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item2', 'item6']
print(list1.index('item2'))
```

```python
# count
list1 = ['item1', 'item2', 'item3', 'item4', 'item5', 'item2', 'item6']
print(list1.count('item2'))
```

```python
# sort
list1 = ['item1', 'item4', 'item2', 'item5', 'item2', 'item6', 'item3']
list1.sort(reverse=True)
print(list1)
```

```python
# reverse
list1 = ['item1', 'item4', 'item2', 'item5', 'item2', 'item6', 'item3']
list1.reverse()
print(list1)
```

```python
# copy
list1 = ['item1', 'item4', 'item2', 'item5', 'item2', 'item6', 'item3']
list2 = list1.copy()
print(list2)
```

```python
# Sum
list1 = [i for i in range(1,11)]
print(list1)
print(sum(list1))
```

```python
# Min
list1 = [i for i in range(1,11)]
print(list1)
print(min(list1))
```

```python
# Max
list1 = [i for i in range(1,11)]
print(list1)
print(max(list1))
```
