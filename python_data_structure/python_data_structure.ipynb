#Python Data Structure

A data structure is simply a way of organizing data so that we can work with it efficiently.
<br>
Python Data Structure:
* List — ordered, mutable collections
* Tuple — ordered, immutable collections
* Set — unique values + fast membership testing
* Dictionary — key-value data; extremely important for DE

## List

* ordered collection of items
* mutable
* square brackets ([])
<br>
Note:
<br>

*   Python automatically adjusts out-of-range slice boundaries. No IndexError is raised.



### Creating a List

```python
#declaring an empty list

my_list1 = []
print(my_list1)
```

```python
# A list of mixed data types
my_list = [1, "Python", 3.14]
print(my_list)
```

### Accessing List Items

```python
my_list = ["item"+str(i) for i in range(1,10)]
print(my_list)
```

```python
#Accessing the first item
print(my_list[0])

#Accessing the forth item
print(my_list[3])
```

```python
#Accessing the first item
print(my_list[-1])

#Accessing the third last item
print(my_list[-3])
```

### Modifying elements in a list

list[index] = new_value

### List Slice

list[begin: end: step] <br>
being - starting index <br>
end - boundary index (is not included) <br>
step -  determines how many elements the slice "steps" forward or backward
<br>
<br>
image.png

```python
my_list = ['item'+str(i) for i in range(1,10)]
print(my_list)
```

```python
# First n elements
print(my_list[:3])
```

```python
# Last n elements
print(my_list[-3:])
```

```python
# Every nth element from a list
print(my_list[::2])
```

```python
# Reversing a list
print(my_list[::-1])
```

```python
# Slicing a list
print(my_list[2:6])
print(my_list[-4:-2])
print(my_list[2:8:2])
print(my_list[-1::-2])
```

### Converting to a list

```python
# Converting an iterable (such as strings, range() etc.) into a list

# Coverts a string into a list
vowels_list = list('aeiou')
print(vowels_list)

# Converts a range into a list
numbers_list = list(range(1, 6))
print(numbers_list)
```
