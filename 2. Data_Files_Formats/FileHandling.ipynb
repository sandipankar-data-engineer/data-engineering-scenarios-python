# File Handling

File handling refers to the process of performing operations on a file, such as creating, opening, reading, writing and closing it through a programming interface.
<br>
<br>
Need for File Handling:
* Store data permanently, even after the program ends.
* Access external files like .txt, .csv, .json, etc.
* Process large files efficiently without using much memory.
* Automate tasks like reading configs or saving outputs.

There are four different modes for opening a file:
* "r" - Read. Default Value. Opens a file for reading. Error if the file doesn't exist.
* "a" - Append. Opens a file for appending, creates the file if it does not exist.
* "w" - Write. Opens a file for writing, creates the file if it doesn't exist.
* "x" - Create. Creates the specified file, returns an error if the file exists.

Below are file modes: 
* "t" - Text - Default value. Text mode
* "b" - Binary - Binary mode (e.g. images)

Syntax:
<br>
file = open("filename", "modes")

## Read Files

### Open a File on the Server

```python
file = open("README.md", 'rt')
print(file.read())
```

### Using the with statement
open() function returns a file object<br>
read() method returns the whole text

```python
with open("README.md", 'rt') as file:
  print(file.read())
```

### Close Files
It is a good practice to always close the file when you are done with it.

```python
file = open("README.md", 'rt')
print(file.read())
file.close()
```

### Read Only Parts of the File
read() method returns the whole text, but you can also specify how many characters you want to return

```python
with open("README.md") as file:
  print(file.read(5)) #Returns the first 5 characters from the file
```

### Read Lines
readline()/readlines() function returns only line

```python
with open("README.md") as file:
  print(file.readline())
  print(file.readline())
```

### Reading a File Using a Loop

```python
file = open("README.md", "r")
for line in file:
    print(line)
file.close()
```

### Check File Pointer

```python
with open("notes.txt", "r") as file:
    print(file.tell())
    file.read(5)
    print(file.tell())
```

### Move the File Pointer

```python
with open("notes.txt", "r") as file:
    print(file.read(12))
    file.seek(12)
    print(file.read())
    file.close()
```

## Write Files
To write to an existing file, you must add a parameter.<br>
write() method takes a single string argument and writes it to the file.

```python
file_copy = open("README.md","rt")
file_paste = open("README_copy.md", "wt")
file_paste.write(file_copy.read())
file_copy.close()
file_paste.close()
```

```python
# "w" creates a file if doesnt exist, completely erases any exisiting content
with open("Dummy.txt", "wt") as file:
    file.write("This is a dummy file with dummy text")
    file.close()

with open("Dummy.txt", "rt") as file:
    print(file.read())
    file.close()
```

```python
# 'a' creates a file if doenst exist, append at the end of exisiting content
with open("Dummy.txt", "at") as file:
    file.write("This is a dummy file with dummy text 2")
    file.close()

with open("Dummy.txt", "rt") as file:
    print(file.read())
    file.close
```

```python
# throughs error if file exist, just to create am empty file
file = open("Dummy.txt", "x")
file.close()
```

### Writing Multiple Lines

```python
file = open("notes.txt", "w")
file.write("Python\n")
file.write("SQL\n")
file.write("Spark\n")
file.close()
```

### Writelines Function

```python
lines = [
    "Python\n",
    "SQL\n",
    "Spark\n"
]
with open("notes.txt", "w") as file:
    file.writelines(lines)
```

## Other Modes
* r+ - Existing file required. Existing content preserved. Write where current pointer is preserved. During Read Initial Pointer is at the Beginning. **Overwrite**.
* w+ - Existing file not required. Existing content is erased. Write where current pointer is preserved. During Read Initial Pointer is at the Beginning. **Erase**. 
* a+ - Exisiting file not required. Existing content preserved. Write always at the end pf the file. During Read Initial Pointer is at the End of the file. **Append**.
